### 3. Connexion Python

````python
from pymongo import MongoClient
from bson.json_util import loads

client = MongoClient("mongodb://localhost:27017/")
db = client["ecommerce"]
````

### 4. Import des clients

````python
with open("data/clients.json", encoding="utf-8") as fichier:
    clients = loads(fichier.read())

db.clients.delete_many({})
db.clients.insert_many(clients)
````

### 5. Import des produits

````python
with open("data/produits.json", encoding="utf-8") as fichier:
    produits = loads(fichier.read())

db.produits.delete_many({})
db.produits.insert_many(produits)
````

### 6. Import des commandes

````python
with open("data/commandes.json", encoding="utf-8") as fichier:
    commandes = loads(fichier.read())

db.commandes.delete_many({})
db.commandes.insert_many(commandes)
````

### 7. Import des ventes

````python
with open("data/ventes.json", encoding="utf-8") as fichier:
    ventes = loads(fichier.read())

db.ventes.delete_many({})
db.ventes.insert_many(ventes)
````

### 8. insertOne() / insert_one()

````python

db.atelier_crud.insert_one({
    "_id": "D1",
    "nom": "Ali",
    "age": 22
})
````

### 9. insertMany() / insert_many()

````python
db.atelier_crud.insert_many([
    {"_id": "D2", "nom": "Sara", "age": 25, "ville": "Rabat"},
    {"_id": "D3", "nom": "Omar", "age": 30, "telephone": "0600000000"}
])
````

### 10.find_one()

### 11. find()

````python
list(db.atelier_crud.find())
````

### 12. updateOne() / update_one()

````python
db.atelier_crud.update_one(
    {"_id": "D1"},
    {"$set": {"age": 23}}
)
````

### 13. updateMany() / update_many()

````python
db.atelier_crud.update_many(
    {"_id": {"$in": ["D2", "D3"]}},
    {"$inc": {"age": 1}}
)
````

### 14.delete_one()

````python
db.atelier_crud.delete_one({"_id": "D1"})
````

### 15. deleteMany() / delete_many()

````python
db.atelier_crud.delete_many({
    "_id": {"$in": ["D2", "D3"]}
})
````

### 16. Opérateurs `$gt`, `$gte`, `$lt`, `$lte`

````python
list(db.produits.find({
    "prix": {"$gte": 100, "$lte": 250}
}))
````

### 17. Opérateurs `$in`, `$ne`

````python
list(db.produits.find({
    "categorie": {"$in": ["Mode", "Maison"]},
    "prix": {"$ne": 200}
}))
````

### 18. Opérateurs `$and`, `$or`

````python
list(db.produits.find({
    "$and": [
        {"prix": {"$gt": 100}},
        {"$or": [
            {"categorie": "Mode"},
            {"categorie": "Électronique"}
        ]}
    ]
}))
````

### 19. Opérateur `$exists`

````python
list(db.clients.find({
    "telephone": {"$exists": True}
}))
````

## Partie 2 — Votre ami (10 minutes)

### 20. `$match`

````python
list(db.ventes.aggregate([
    {"$match": {"categorie": "Mode"}},
    {"$limit": 2}
]))
````

### 21. Produits les plus vendus — `$group`, `$sort`, `$limit`, `$lookup`, `$project`

````python
list(db.ventes.aggregate([
    {"$group": {
        "_id": "$produit_id",
        "quantite": {"$sum": "$quantite"}
    }},
    {"$sort": {"quantite": -1}},
    {"$limit": 5},
    {"$lookup": {
        "from": "produits",
        "localField": "_id",
        "foreignField": "_id",
        "as": "produit"
    }},
    {"$project": {
        "_id": 0,
        "produit": "$produit.nom",
        "quantite":1,
    }}
]))

````

### 22. Chiffre d'affaires par catégorie — `$group`, `$sort`

````python
ca_categories = list(db.ventes.aggregate([
    {"$group": {
        "_id": "$categorie",
        "chiffre_affaires": {
            "$sum": {"$multiply": ["$quantite", "$prix"]}
        }
    }},
    {"$sort": {"chiffre_affaires": -1}}
]))

ca_categories
````

### 23. Meilleurs clients — `$group`, `$sort`, `$lookup`, `$project`

````python
meilleurs_clients = list(db.ventes.aggregate([
    {"$group": {
        "_id": "$client_id",
        "total": {
            "$sum": {"$multiply": ["$quantite", "$prix"]}
        }
    }},
    {"$sort": {"total": -1}},
    {"$lookup": {
        "from": "clients",
        "localField": "_id",
        "foreignField": "_id",
        "as": "client"
    }},
    {"$project": {
        "_id": 0,
        "client": "$client.nom",
        "total": 1
    }}
]))

meilleurs_clients
````

### 24. Ventes par mois — `$group`, `$sort`, `$project`

````python
ventes_par_mois = list(db.ventes.aggregate([
    {"$group": {
        "_id": {
            "$dateToString": {
                "format": "%Y-%m",
                "date": "$date"
            }
        },
        "chiffre_affaires": {
            "$sum": {"$multiply": ["$quantite", "$prix"]}
        },
        "quantite": {"$sum": "$quantite"}
    }},
    {"$sort": {"_id": 1}},
    {"$project": {
        "_id": 0,
        "mois": "$_id",
        "chiffre_affaires": 1,
        "quantite": 1
    }}
]))

ventes_par_mois
````
